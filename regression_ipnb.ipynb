```python
"""
1. What is Simple Linear Regression?

Simple Linear Regression is a method used to find the relationship between one independent variable (X) and one dependent variable (Y) using a straight line.



2. Key assumptions of Simple Linear Regression
Linear relationship between X and Y
Errors are normally distributed
Constant variance (no heteroscedasticity)
Observations are independent



3. What does coefficient m represent in Y = mX + c?

The coefficient m represents the slope, i.e., how much Y changes when X increases by 1 unit.




4. What does intercept c represent?

The intercept c is the value of Y when X = 0.




5. How do we calculate slope m?

Slope is calculated using:

𝑚=∑(𝑋−𝑋ˉ)(𝑌−𝑌ˉ)/∑(𝑋−𝑋ˉ)2


	​

6. Purpose of Least Squares Method

It minimizes the sum of squared errors between actual and predicted values to find the best fit line.




7. Interpretation of R² (coefficient of determination)

R² tells how much variation in Y is explained by X.

R² = 1 → perfect fit
R² = 0 → no relationship



8. What is Multiple Linear Regression?

It is regression with more than one independent variable predicting a dependent variable.




9. Difference: Simple vs Multiple Regression
Simple: 1 independent variable
Multiple: 2 or more independent variables



10. Key assumptions of Multiple Linear Regression
Linear relationship
No multicollinearity
Homoscedasticity
Normal distribution of errors
Independence of observations




11. What is heteroscedasticity?

It means unequal variance of errors.
It can make predictions less reliable.




12. How to improve multicollinearity?
Remove correlated variables
Use PCA
Apply regularization (Ridge/Lasso)




13. Techniques for categorical variables
One-hot encoding
Label encoding
Dummy variables




14. Role of interaction terms

They show how two variables together affect Y.





15. Intercept interpretation difference
Simple: Y when X = 0
Multiple: Y when all X variables = 0




16. Significance of slope

Slope shows direction and strength of relationship and affects prediction.





17. Role of intercept in context

Intercept gives a baseline value for prediction.





18. Limitations of R²
Does not show causation
Can increase with more variables
Ignores overfitting




19. Large standard error meaning

It means the coefficient estimate is not reliable.




20. Identifying heteroscedasticity
Seen in residual plots (unequal spread)
Important because it affects accuracy





21. High R² but low Adjusted R²

It means irrelevant variables are added.




22. Why scale variables?
Improves model performance
Helps in faster convergence
Important for algorithms like gradient descent




23. What is Polynomial Regression?

It models relationships using curved lines instead of straight lines.





24. Difference: Polynomial vs Linear
Linear → straight line
Polynomial → curved line





25. When is Polynomial Regression used?

When data shows a non-linear relationship.





26. General equation
𝑌=𝑎+𝑏1𝑋+𝑏2𝑋^2+𝑏3𝑋^3+...



27. Can it be applied to multiple variables?

Yes, it can be extended to multiple variables.




28. Limitations of Polynomial Regression
Overfitting
Complex interpretation
Sensitive to outliers




29. Methods to select degree
Cross-validation
Adjusted R²
Mean Squared Error (MSE)



30. Why visualization is important?

It helps to understand data pattern and choose correct degree.
"""



#31. Implementation in Python:
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample data (independent variable)
X = np.array([[1], [2], [3], [4], [5]])

# Dependent variable
y = np.array([2, 4, 5, 4, 5])

# Convert to polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Train model
model = LinearRegression()
model.fit(X_poly, y)

# Prediction
y_pred = model.predict(X_poly)

print(y_pred)
```
