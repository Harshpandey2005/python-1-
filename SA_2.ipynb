```python
"""
Q1. Z-statistic

The z-statistic (z-score) measures how many standard deviations a data point is away from the mean.

Formula:


z=X−μ/σ
	​

It is related to the standard normal distribution, which has mean = 0 and standard deviation = 1.
In hypothesis testing, z-score helps us decide whether to reject the null hypothesis by comparing it with critical values."""
```

```python
"""
Q2. P-value

The p-value is the probability of getting results as extreme as the observed result, assuming the null hypothesis is true.

If p-value is small (e.g., 0.01) → strong evidence against null hypothesis.
If p-value < significance level (0.05) → reject null hypothesis."""
```

```python
"""Q9. Calculation (Important)

Given:

Sample mean = 105
Population mean = 100
SD = 15
n = 25
Step 1: Z-score
z=105−100/15/5
=5/3=1.67
Step 2: P-value

From z-table:
p ≈ 0.095 (two-tailed)

Step 3: Decision
p (0.095) > 0.05
👉 Fail to reject null hypothesis"""


#PYTHON CODE:
import scipy.stats as stats
import numpy as np

# Given data
sample_mean = 105
population_mean = 100
std_dev = 15
n = 25

# Calculate z-score
z = (sample_mean - population_mean) / (std_dev / np.sqrt(n))

# Calculate p-value (two-tailed)
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Z-score:", z)
print("P-value:", p_value)
```

```python
"""Q10. Binomial Distribution Simulation
Expected:
Mean = n × p = 10 × 0.6 = 6
Variance = n × p × (1-p) = 10 × 0.6 × 0.4 = 2.4"""


#Python Code:
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10
p = 0.6
samples = 1000

# Generate data
data = np.random.binomial(n, p, samples)

# Plot
plt.hist(data, bins=10)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()

# Mean and variance
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
```
