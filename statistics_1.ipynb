```python
"""
1. Types of Data

Data is information collected for analysis.

(a) Qualitative Data (Categorical)
Describes qualities or categories
Example: Gender, Color, City
(b) Quantitative Data (Numerical)
Represents numbers
Example: Height, Weight, Marks
Scales of Measurement:
Nominal: No order (e.g., colors)
Ordinal: Has order (e.g., ranks)
Interval: Equal intervals, no true zero (e.g., temperature in °C)
Ratio: Has true zero (e.g., height, weight)



2. Measures of Central Tendency

They show the center of data.

Mean (Average)
Sum of values ÷ total values
Use when data has no extreme values
Median
Middle value
Use when data has outliers
Mode
Most frequent value
Use for categorical data




3. Dispersion

Dispersion shows how spread out data is.

Variance
Average of squared differences from mean
Standard Deviation
Square root of variance
Easy to understand spread

 Higher value = more spread




4. Box Plot

A box plot shows:

Minimum
Q1 (25%)
Median
Q3 (75%)
Maximum

 Helps identify:

Spread
Skewness
Outliers




5. Random Sampling
Selecting data randomly from a population
Ensures fair representation

 Important for:

Accurate conclusions
Avoiding bias



6. Skewness

Skewness shows data symmetry.

Types:
Positive skew: Tail on right (mean > median)
Negative skew: Tail on left (mean < median)
Zero skew: Symmetrical

 Affects interpretation of average




7. Interquartile Range (IQR)
𝐼𝑄𝑅=𝑄3−𝑄1


 Used to detect outliers:

Lower limit = Q1 − 1.5 × IQR
Upper limit = Q3 + 1.5 × IQR

Values outside = outliers




8. Binomial Distribution Conditions

Used when:

Fixed number of trials
Only 2 outcomes (success/failure)
Same probability each time
Trials are independent




9. Normal Distribution
Bell-shaped curve
Symmetrical
Mean = Median = Mode
Empirical Rule:
68% → within 1 SD
95% → within 2 SD
99.7% → within 3 SD



10. Poisson Process Example

Example: Number of calls in 1 minute

If average = 3 calls/min
Probability of 2 calls:

𝑃(𝑋=2)=𝑒^−3.3^2 /2!
Used for rare events over time



11. Random Variable

A variable whose value depends on chance.

Types:
Discrete: Countable (e.g., number of students)
Continuous: Measurable (e.g., height, time)



12. Covariance & Correlation

Example dataset:

X = 1, 2, 3
Y = 2, 4, 6

Covariance
Shows direction of relationship
Positive → move together
Correlation (r)
Value between -1 to +1
+1 = strong positive relation

 Interpretation:
As X increases, Y also increases → strong positive relation"""
```
