```python
"""1. Properties of F-distribution
It is always positive (≥ 0).
It is right-skewed (not symmetrical).
It depends on two degrees of freedom (df₁ and df₂).
Shape changes with different degrees of freedom.
Used to compare variances of two samples.





2. Where F-distribution is used & why

Used in:

F-test (compare two variances)
ANOVA (Analysis of Variance)

👉 Why appropriate:

Because it is based on the ratio of two variances, and F-distribution describes this ratio perfectly.



3. Assumptions of F-test
Data should be normally distributed
Samples must be independent
Data should be randomly selected
Populations should have equal means (for variance comparison)




4. Purpose of ANOVA & difference from t-test
ANOVA: compares means of 3 or more groups
t-test: compares means of only 2 groups

👉 Difference:

t-test → 2 groups
ANOVA → 3+ groups
ANOVA reduces error compared to multiple t-tests




5. Why use one-way ANOVA instead of multiple t-tests
Multiple t-tests increase error (Type I error)
ANOVA tests all groups at once
More accurate and efficient



6. Variance partition in ANOVA

Total variance is divided into:

Between-group variance → difference between group means
Within-group variance → variation inside each group

👉 F-statistic formula:

F=Within-group variance/Between-group variance
	​


👉 If between-group variance is high → groups are different.




7. Classical vs Bayesian ANOVA
Feature	            Classical (Frequentist)	     Bayesian
Uncertainty       	Uses probability of data     Uses probability of parameters
Parameters	        Fixed	                       Treated as random
Result	            p-value                    	 Posterior probability
Interpretation    	Reject / Accept hypothesis	 Degree of belief"""
```

```python
"""8. F-test (Profession A vs B)
Data:
A: [48, 52, 55, 60, 62]
B: [45, 50, 55, 52, 47]
Python Code:"""


import numpy as np
from scipy.stats import f

A = np.array([48, 52, 55, 60, 62])
B = np.array([45, 50, 55, 52, 47])

var_A = np.var(A, ddof=1)
var_B = np.var(B, ddof=1)

F = var_A / var_B

df1 = len(A) - 1
df2 = len(B) - 1

p_value = 1 - f.cdf(F, df1, df2)

print("F-statistic:", F)
print("p-value:", p_value)

"""
Conclusion:
If p > 0.05 → variances are equal
If p < 0.05 → variances are different

👉 Here, p-value will be greater than 0.05, so
✅ Variances are equal"""
```

```python
"""9. One-way ANOVA (Regions)
Data:
A: [160, 162, 165, 158, 164]
B: [172, 175, 170, 168, 174]
C: [180, 182, 179, 185, 183]
Python Code:"""

from scipy.stats import f_oneway

A = [160, 162, 165, 158, 164]
B = [172, 175, 170, 168, 174]
C = [180, 182, 179, 185, 183]

F, p = f_oneway(A, B, C)

print("F-statistic:", F)
print("p-value:", p)

"""Conclusion:
p-value will be < 0.05
👉 So,
✅ There is a significant difference in heights between regions"""
```
