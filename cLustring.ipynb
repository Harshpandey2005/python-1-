```python
"""
THEORETICAL QUESTIONS
1. What is unsupervised learning in machine learning?

Unsupervised learning is a type of machine learning where the model works on unlabeled data. It tries to find hidden patterns, structures, or groups in the dataset without predefined outputs.

2. How does K-Means clustering algorithm work?

K-Means divides data into K clusters. It randomly selects K centroids, assigns points to nearest centroid, updates centroids, and repeats until centroids stop changing.

3. Explain dendrogram in hierarchical clustering.

A dendrogram is a tree-like diagram used in hierarchical clustering. It shows how clusters are merged or divided at different distances.

4. Difference between K-Means and Hierarchical Clustering.

K-Means requires number of clusters beforehand and forms flat clusters. Hierarchical clustering creates a tree of clusters and does not always need predefined K.

5. Advantages of DBSCAN over K-Means.

DBSCAN can find clusters of any shape, does not require number of clusters beforehand, and can identify noise/outliers.

6. When would you use Silhouette Score?

Silhouette Score is used to evaluate clustering quality by measuring how similar a point is to its own cluster compared to other clusters.

7. Limitations of Hierarchical Clustering.

It is slow for large datasets, sensitive to noise, and once clusters merge they cannot be undone.

8. Why is feature scaling important in clustering?

Feature scaling ensures all features contribute equally. Without scaling, larger-value features dominate distance calculations.

9. How does DBSCAN identify noise points?

Points that do not belong to any dense region and have fewer neighbors than minimum samples are marked as noise.

10. Define inertia in K-Means.

Inertia is the sum of squared distances of data points from their nearest cluster centroid.

11. What is elbow method in K-Means?

Elbow method helps choose optimal K by plotting inertia vs K. The bend point is selected.

12. Describe density in DBSCAN.

Density means number of points within a given radius around a point.

13. Can hierarchical clustering be used on categorical data?

Yes, if suitable distance measures such as Hamming distance are used.

14. What does negative Silhouette Score indicate?

It indicates points may be assigned to wrong clusters.

15. Explain linkage criteria in hierarchical clustering.

Linkage criteria decide distance between clusters such as single, complete, average linkage.

16. Why K-Means performs poorly on varying cluster sizes?

Because it assumes spherical and equal-sized clusters.

17. Core parameters in DBSCAN.

eps = neighborhood radius
min_samples = minimum points required for dense region

18. How does K-Means++ improve K-Means?

It selects better initial centroids to improve accuracy and convergence.

19. What is agglomerative clustering?

It is bottom-up hierarchical clustering where each point starts as cluster and merges step by step.

20. Why Silhouette better than inertia?

Silhouette considers both cluster compactness and separation, while inertia only measures compactness."""
```

```python
#PRACTICAL QUESTIONS



#1. Generate synthetic data with 4 centers using make_blobs and apply KMeans.
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

X, y = make_blobs(n_samples=300, centers=4, random_state=42)

kmeans = KMeans(n_clusters=4, random_state=42)
labels = kmeans.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.show()




#2. Load Iris dataset and Agglomerative Clustering.
from sklearn.datasets import load_iris
from sklearn.cluster import AgglomerativeClustering

iris = load_iris()
X = iris.data

model = AgglomerativeClustering(n_clusters=3)
labels = model.fit_predict(X)

print(labels[:10])



#3. make_moons and DBSCAN.
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

X, y = make_moons(n_samples=300, noise=0.1)

db = DBSCAN(eps=0.2, min_samples=5)
labels = db.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.show()




#4. Wine dataset with KMeans.
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import numpy as np

wine = load_wine()
X = wine.data

X = StandardScaler().fit_transform(X)

model = KMeans(n_clusters=3)
labels = model.fit_predict(X)

print(np.bincount(labels))



#5. make_circles with DBSCAN.
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

X, y = make_circles(n_samples=300, noise=0.05)

model = DBSCAN(eps=0.2, min_samples=5)
labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.show()




#6. Breast Cancer + MinMaxScaler + KMeans
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans

data = load_breast_cancer()
X = data.data

X = MinMaxScaler().fit_transform(X)

model = KMeans(n_clusters=2)
model.fit(X)

print(model.cluster_centers_)




#7. make_blobs varying std + DBSCAN
X, y = make_blobs(n_samples=300, centers=3, cluster_std=[1.0,2.5,0.5])

model = DBSCAN(eps=0.8)
labels = model.fit_predict(X)




#8. Digits + PCA + KMeans
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA

digits = load_digits()
X = digits.data

X_pca = PCA(n_components=2).fit_transform(X)

model = KMeans(n_clusters=10)
labels = model.fit_predict(X_pca)

plt.scatter(X_pca[:,0], X_pca[:,1], c=labels)
plt.show()



#9. Silhouette score for k=2 to 5
from sklearn.metrics import silhouette_score

scores = []
for k in range(2,6):
    model = KMeans(n_clusters=k)
    labels = model.fit_predict(X)
    scores.append(silhouette_score(X, labels))

plt.bar(range(2,6), scores)
plt.show()



#10. Iris dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage

Z = linkage(X, method='average')

dendrogram(Z)
plt.show()



#11. Overlapping blobs + KMeans
X, y = make_blobs(n_samples=300, centers=3, cluster_std=2.5)

model = KMeans(n_clusters=3)
labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.show()



#12. Digits + tSNE + DBSCAN
from sklearn.manifold import TSNE

X_2d = TSNE(n_components=2).fit_transform(X)

labels = DBSCAN(eps=5).fit_predict(X_2d)




#13. Agglomerative complete linkage
model = AgglomerativeClustering(n_clusters=3, linkage='complete')
labels = model.fit_predict(X)



#14. Breast Cancer inertia K=2 to 6
inertia = []

for k in range(2,7):
    model = KMeans(n_clusters=k)
    model.fit(X)
    inertia.append(model.inertia_)

plt.plot(range(2,7), inertia)
plt.show()



#15. Circles + Agglomerative single linkage
model = AgglomerativeClustering(n_clusters=2, linkage='single')
labels = model.fit_predict(X)


#16. Wine + DBSCAN cluster count
labels = DBSCAN(eps=1).fit_predict(X)

print(len(set(labels)) - (1 if -1 in labels else 0))



#17. make_blobs + centers plot
model = KMeans(n_clusters=3)
model.fit(X)

plt.scatter(X[:,0], X[:,1])
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], s=200)
plt.show()



#18. Iris + DBSCAN noise count
labels = DBSCAN(eps=0.5).fit_predict(X)

print(sum(labels == -1))



#19. make_moons + KMeans
X, y = make_moons(n_samples=300, noise=0.1)

labels = KMeans(n_clusters=2).fit_predict(X)



#20. Digits + PCA 3D + KMeans
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load dataset
digits = load_digits()
X = digits.data

# Reduce to 3 components
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)

# Apply KMeans
kmeans = KMeans(n_clusters=10, random_state=42)
labels = kmeans.fit_predict(X_pca)

# 3D Scatter Plot
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")

plt.show()
```

```python
#21. Blobs with 5 centers + KMeans + silhouette score
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, y = make_blobs(n_samples=300, centers=5, random_state=42)

model = KMeans(n_clusters=5)
labels = model.fit_predict(X)

score = silhouette_score(X, labels)
print(score)




#22. Breast Cancer + PCA + Agglomerative
from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

data = load_breast_cancer()
X = data.data

X_pca = PCA(n_components=2).fit_transform(X)

labels = AgglomerativeClustering(n_clusters=2).fit_predict(X_pca)

plt.scatter(X_pca[:,0], X_pca[:,1], c=labels)
plt.show()




#23. Noisy circles + KMeans and DBSCAN side by side
from sklearn.datasets import make_circles
from sklearn.cluster import KMeans, DBSCAN
import matplotlib.pyplot as plt

X, y = make_circles(n_samples=300, noise=0.1)

k_labels = KMeans(n_clusters=2).fit_predict(X)
d_labels = DBSCAN(eps=0.2).fit_predict(X)

plt.subplot(1,2,1)
plt.scatter(X[:,0], X[:,1], c=k_labels)

plt.subplot(1,2,2)
plt.scatter(X[:,0], X[:,1], c=d_labels)

plt.show()






#24. Iris + Silhouette Coefficient for each sample
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples
import matplotlib.pyplot as plt

iris = load_iris()
X = iris.data

labels = KMeans(n_clusters=3).fit_predict(X)

scores = silhouette_samples(X, labels)

plt.bar(range(len(scores)), scores)
plt.show()






#25. Blobs + Agglomerative average linkage
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

X, y = make_blobs(n_samples=300, centers=3)

labels = AgglomerativeClustering(n_clusters=3, linkage='average').fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.show()





#26. Wine + KMeans + pairplot
from sklearn.datasets import load_wine
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans

wine = load_wine()

df = pd.DataFrame(wine.data[:, :4], columns=wine.feature_names[:4])

labels = KMeans(n_clusters=3).fit_predict(df)

df["Cluster"] = labels

sns.pairplot(df, hue="Cluster")




#27. Noisy blobs + DBSCAN cluster and noise count
from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
import numpy as np

X, y = make_blobs(n_samples=300, centers=3, cluster_std=1.5)

labels = DBSCAN(eps=0.5).fit_predict(X)

print("Noise Points:", sum(labels == -1))
print("Clusters:", len(set(labels)) - (1 if -1 in labels else 0))





#28. Digits + t-SNE + Agglomerative
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

digits = load_digits()
X = digits.data

X_2d = TSNE(n_components=2).fit_transform(X)

labels = AgglomerativeClustering(n_clusters=10).fit_predict(X_2d)

plt.scatter(X_2d[:,0], X_2d[:,1], c=labels)
plt.show()
```
