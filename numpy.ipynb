```python
#9.
arr = np.random.randint(0, 1000, 100)

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

primes = [x for x in arr if is_prime(x)]
print(primes)
```

```python
#10.
temps = np.random.randint(20, 40, 30)

weeks = np.array_split(temps, 4)

for i, week in enumerate(weeks):
    print(f"Week {i+1} avg:", np.mean(week))
```
